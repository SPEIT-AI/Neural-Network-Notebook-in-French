# GRU
Le GRU (Gated Recurrent Unit) est une variante simplifiée du LSTM (Long Short-Term Memory), introduite par Cho et al. en 2014. Il vise à résoudre le problème du gradient qui disparaît dans les réseaux neuronaux récurrents classiques, tout en étant plus léger en paramètres que le LSTM.

Contrairement au LSTM qui utilise trois portes (entrée, oubli, sortie), le GRU n’en utilise que deux : la porte de mise à jour (update gate) et la porte de réinitialisation (reset gate).



On utilise les mêmes notations que pour la LSTM :

* $x_t$ : vecteur d'entrée à l'instant $t$

* $h_t$ : état caché à l'instant $t$

* $y_t$ : sortie prédite à l’instant $t$

* $W, b$ : paramètres appris (poids, biais)

* $\sigma$ : fonction sigmoïde, $\tanh$ : tangente hyperbolique

## 1 - Réseau à Unités Récurrentes à Portes (GRU)

Le GRU simplifie l'architecture LSTM tout en conservant un bon pouvoir de mémorisation. Il ne contient pas d'état de cellule séparé $c_t$ : tout est contenu dans l'état caché $h_t$. On y trouve deux portes principales :

<img src="GRU_mots.png" alt="image_description" width="800">

<caption><center> Figure 1 : Modèle basique de RNN</center></caption>

#### - Porte de mise à jour (update gate)
Cette porte décide combien de l'ancien état $h_{t-1}$ on conserve :
$$(\Gamma_u)_t = \sigma_u\left(w_u^{\top} \cdot x_t + v_u^{\top} \cdot h_{t-1} + b_u\right) \tag{1}$$

#### - Porte de réinitialisation (reset gate)
Cette porte décide combien de l'ancien état $h_{t-1}$ est ignoré lors de la création du nouveau contenu :
$$(\Gamma_r)_t = \sigma_r\left(w_r^{\top} \cdot x_t + v_r^{\top} \cdot h_{t-1} + b_r\right) \tag{2}$$

#### - État candidat
On calcule un nouvel état candidat $\tilde{h}_t$ :
$$\tilde{h}_t = \tanh\left(w_h^{\top} \cdot x_t + v_h^{\top} \cdot \left((\Gamma_r)_t \odot h_{t-1}\right) + b_h\right) \tag{3}$$

#### - État caché final
On interpole entre l’ancien état et l’état candidat avec $(\Gamma_u)_t$ :
$$h_t = (\Gamma_u)_t \odot h_{t-1} + (1-(\Gamma_u)_t) \odot \tilde{h}_t  \tag{4}$$

#### - Sortie
Enfin, on calcule la sortie prédite :
$$\hat{y}_t = \text{softmax}(u_y^{\top} \cdot h_t + b_y) \tag{5}$$

### 1.3 - Rétropropagation pour un réseau GRU

La rétropropagation pour un réseau GRU consiste à calculer les gradients de la fonction de coût par rapport aux paramètres du réseau. Cela se fait en parcourant la séquence en sens inverse et en appliquant la règle de la chaîne.

On suppose que l’on minimise une fonction de perte $L$ dépendant de $h_t$. Notre but est de calculer les gradients
$$\frac{\partial L}{\partial W_u}, \frac{\partial L}{\partial W_r}, \frac{\partial L}{\partial W_h}$$

**Étape 1 : Gradient de la perte par rapport à $h_t$**
C’est le signal de départ
$$\mathrm{d} h_t = \frac{\partial L}{\partial h_t} \tag{6}$$

**Étape 2 : Gradient par rapport aux composantes internes**
- Gradient par rapport à $\tilde{h}_t$
$$\mathrm{d} \tilde{h}_t = \mathrm{d} h_t \odot (\Gamma_u)_t \tag{7}$$

- Puis, dérivée de $\tanh$
$$\mathrm{d} \tanh = \mathrm{d} \tilde{h}_t \odot (1-\tilde{h}_t^2) \tag{8}$$

- Gradient par rapport à $(\Gamma_u)_t$
$$\mathrm{d} (\Gamma_u)_t = \mathrm{d} h_t \odot (h_{t-1}-\tilde{h}_t) \tag{9}$$

- Puis, dérivée de la sigmoïde $\sigma_u$
$$\mathrm{d} \sigma_u = \mathrm{d} (\Gamma_u)_t \odot (\Gamma_u)_t \odot (1-(\Gamma_u)_t) \tag{10}$$

- Gradient par rapport à $(\Gamma_r)_t$ (à travers $\tilde{h}_t$)
$$\mathrm{d} (\Gamma_r)_t = \left(\mathrm{d} \tanh \cdot v_h\right) \odot h_{t-1} \tag{11}$$
$$\mathrm{d} \sigma_r = \mathrm{d} (\Gamma_r)_t \odot (\Gamma_r)_t \odot (1-(\Gamma_r)_t) \tag{12}$$

**Étape 3 : Rétropropagation aux poids**
Les dérivées par rapport aux poids (ex : $w_u$ et $v_u$)
$$\frac{\partial L}{\partial w_u} = \mathrm{d} \sigma_u \cdot x_t^T \tag{13}$$
$$\frac{\partial L}{\partial v_u} = \mathrm{d} \sigma_u \cdot h_{t-1}^T \tag{14}$$
$$\frac{\partial L}{\partial b_u} = \mathrm{d} \sigma_u \tag{15}$$

3. Gradients des poids et biais : Les gradients des matrices de poids (`dWu`, `dWr`, `dWh`) et des biais (`dbu`, `dbr`, `dbh`) sont calculés en utilisant les gradients des portes et les entrées concaténées.

## 2 - Comparaison entre RNN, LSTM et GRU pour le traitement de la langue française (à lire)

Dans cette section, nous allons comparer les performances des réseaux de neurones récurrents traditionnels (RNN), des réseaux de mémoire à long terme (LSTM) et des réseaux à unités récurrentes à portes (GRU) sur une tâche spécifique de traitement de la langue française. Nous utiliserons un exemple concret de prédiction de la prochaine lettre dans une séquence de mots français.

#### - Description de la tâche
La tâche consiste à prédire la prochaine lettre dans une séquence de mots français. Par exemple, si l'entrée est "bonjou", le modèle doit prédire que la prochaine lettre est "r".

#### - Préparation des données
Nous allons utiliser un corpus de mots français pour entraîner nos modèles. Les mots seront tokenisés en caractères, et nous allons créer des séquences d'entrée et de sortie correspondantes.

```python
# Exemple de corpus de mots français
corpus = ["bonjour", "merci", "salut", "au revoir", "s'il vous plaît"]

# Tokenisation des mots en caractères
chars = sorted(list(set("".join(corpus))))
char_to_idx = {char: idx for idx, char in enumerate(chars)}
idx_to_char = {idx: char for idx, char in enumerate(chars)}

# Création des séquences d'entrée et de sortie
def prepare_sequences(corpus, char_to_idx, seq_length):
    X = []
    y = []
    for word in corpus:
        for i in range(len(word) - seq_length):
            seq_in = word[i:i + seq_length]
            seq_out = word[i + seq_length]
            X.append([char_to_idx[char] for char in seq_in])
            y.append(char_to_idx[seq_out])
    return np.array(X), np.array(y)

seq_length = 5
X, y = prepare_sequences(corpus, char_to_idx, seq_length)
print("X.shape =", X.shape)
print("y.shape =", y.shape)
```

### 2.2 - LSTM

- **Structure LSTM** :
    - Le LSTM comporte trois portes principales (oubli, entrée, sortie) et un état mémoire
    - Chaque porte a ses propres paramètres (W, b) qui sont appris pendant l'entraînement
- **Flux de données** :
    - Propagation avant : Calcul séquentiel des états cachés/mémoire et des prédictions
    - Rétropropagation : Calcul précis des gradients à travers les différentes portes
- **Optimisations** :
    - Initialisation soignée des paramètres (valeurs petites et aléatoires)
    - Ajout d'un terme epsilon (1e-8) pour éviter les problèmes numériques
    - Stockage efficace des valeurs intermédiaires pour la rétropropagation
- **Entraînement** :
    - Descentes de gradient avec taux d'apprentissage configurable
    - Affichage périodique de la perte pour suivre la convergence
- **Particularités** :
    - Mécanisme de mémoire à long terme (état c)
    - Portes apprenant quoi oublier/conserver (combinaison non linéaire)
    - Protection contre le vanishing gradient grâce aux connexions résiduelles
